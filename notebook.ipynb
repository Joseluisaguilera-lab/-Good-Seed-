La variable real_age muestra una distribución fuertemente concentrada en adultos jóvenes, especialmente entre los 20 y 25 años, donde se observa el pico más alto del histograma. A partir de los 30–40 años la frecuencia disminuye de manera constante, y en los extremos —edades muy bajas y muy altas— casi no hay datos, lo que evidencia una subrepresentación significativa de niños y adultos mayores. La forma del histograma presenta un sesgo positivo, con una cola larga hacia edades mayores, lo cual coincide con la diferencia entre la media y la mediana. En conjunto, estos hallazgos indican que el dataset está desbalanceado, favoreciendo edades jóvenes y limitando la capacidad del modelo para generalizar correctamente en rangos extremos.

```python
#Configuración del generador con flow_from_dataframe
train_datagen = ImageDataGenerator(rescale=1./255)

train_gen_flow = train_datagen.flow_from_dataframe(
    dataframe=labels,
    directory='/datasets/faces/final_files',
    x_col='file_name',
    y_col='real_age',
    target_size=(224, 224),
    batch_size=32,
    class_mode='raw',
    shuffle=True,
    seed=42
)
```

```python
# Obtener un lote de imágenes y edades
X_batch, y_batch = next(train_gen_flow)

# Crear una cuadrícula de 3 filas x 5 columnas
fig, axes = plt.subplots(3, 5, figsize=(15, 9))

# Recorrer todos los subplots y mostrar cada imagen
for i, ax in enumerate(axes.flat):
    ax.imshow(X_batch[i])
    ax.set_title(f"Edad: {int(y_batch[i])}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

En conjunto, el dataset presenta retos significativos: variabilidad en calidad, ruido en etiquetas y desbalance en la distribución de edades. Define las funciones necesarias para entrenar tu modelo en la plataforma GPU y crea un solo script que las contenga todas junto con la sección de inicialización.

Para facilitar esta tarea, puedes definirlas en este notebook y ejecutar un código listo en la siguiente sección para componer automáticamente el script.

Los revisores del proyecto también verificarán las definiciones a continuación, para que puedan comprender cómo construiste el modelo.

```python
#Codigo de entrenamiento que se ejecuto de manera local
import pandas as pd
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.resnet import ResNet50
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Sequential
from tensorflow.keras.optimizers import Adam

def load_data(path, subset='training'):
    labels = pd.read_csv(path + '\\labels.csv')

    data_datagen = ImageDataGenerator(
        rescale=1./255,
        validation_split=0.2
    )

    data_gen_flow = data_datagen.flow_from_dataframe(
        dataframe=labels,
        directory=path + '\\final_files\\',
        x_col='file_name',
        y_col='real_age',
        target_size=(224, 224),
        batch_size=16,
        class_mode='raw',
        subset=subset,
        seed=12345
    )

    return data_gen_flow


def create_model(input_shape):
    backbone = ResNet50(weights='imagenet', input_shape=input_shape, include_top=False)
    model = Sequential()
    model.add(backbone)
    model.add(GlobalAveragePooling2D())
    model.add(Dense(1, activation='relu'))

    optimizer = Adam(learning_rate=0.0005)
    model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])
    return model


def train_model(model, train_data, test_data, batch_size=None, epochs=3):
    model.fit(train_data, validation_data=test_data, batch_size=batch_size, epochs=epochs, verbose=1)
    return model


# Ruta correcta
data_path = r"C:\Users\luist\OneDrive\Documentos"

input_shape = (224, 224, 3)

train = load_data(data_path, subset='training')
test = load_data(data_path, subset='validation')

model = create_model(input_shape)
model = train_model(model, train, test)
```
